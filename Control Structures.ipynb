```python
# If statement used when is_hot is true
is_hot = True

if is_hot:
    print("It's a hot day")
    print("Drink plenty of water")
print("Enjoy your day")    
```

```python
# If/else control structure for when is_hot
is_hot = True

if is_hot:
    print ("It's a hot day")
    print ("Drink plenty of water")
else:
    print("It's a cold day")
    print("Wear warm clothes")
print("Enjoy your day")
```

```python
# If/else control structure for when is_cold
is_hot = False
is_cold = True

if is_hot:
    print("It's a hot day")
    print("Drink plenty of water")
    
elif is_cold:
    print("It's a cold day")
    print("Wear warm clothes")
    
else:
    print("It's lovely day")
print("Enjoy your day")
```

```python
# If/else control structure with equation for having good credit
price = 1000000
has_good_credit = True

if has_good_credit:
    down_payment = 0.1 * price
else:
    down_payment = 0.2 * price
print(f"Down payment: {down_payment}")
```

```python
# If/else control structure with equation for bad credit
price = 1000000
has_good_credit = False

if has_good_credit:
    down_payment = 0.1 * price
else:
    down_payment = 0.2 * price
print(f"Down payment: ${down_payment}")
```

```python
# If or control structure to show eligible for loan
has_high_income = False
has_good_credit = True

if has_high_income or has_good_credit:
    print("Eligible for loan")
```

```python
# If or control structure to show eligible for loan
has_high_income = True
has_good_credit = False

if has_high_income or has_good_credit:
    print("Eligible for loan")
```

```python
# If or control structure to show no output
has_high_income = False
has_good_credit = False

if has_high_income or has_good_credit:
    print("Eligible for loan")
```

```python
# And not control structure
has_good_credit = True
has_criminal_record = False

if has_good_credit and not has_criminal_record:
    print("Eligible for loan")
```
