```python
# Creates a customer dictionary with key/value pairs
customer = {
    "name": "John Smith",
    "age": 30,
    "is_verified": True
}
```

```python
print(customer["name"]) # prints the value from the name key
```

```python
print(customer.get("name"))
```

```python
print(customer.get("birthdate"))
```

```python
print(customer.get("birthdate", "Jan 1 1980"))
```

```python
customer["name"] = "Jack Smith"
print(customer["name"])
```

```python
customer["birthdate"] = "Jan 1 1980"
print(customer["birthdate"])
```

```python
phone = input("Phone: ") # Takes an input of digits
# Map the digits to values
digits = {
    "1": "One",
    "2": "Two",
    "3": "Three",
    "4": "Four"
}
output = ""
for dig in phone:
    output += digits.get(dig, "!") + " "
print(output) # String output
```
